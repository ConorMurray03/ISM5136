```python
import re
import pandas as pd
from_pattern = r'^From:\s*"?\s*(dr\b\.?[^"<]*?)"?\s*<([\w\.\-+]+@([\w\.\-]+\.(com|net)))>'
names = []
emails = []
domains = []
match_lines = []   
file = open('fraudulent_emails.txt', 'r', encoding='latin-1')
lines = file.readlines()
file.close()
for i in range(len(lines)):
    line = lines[i]
    match = re.search(from_pattern, line, re.IGNORECASE)
    if match:
        names.append(match.group(1).strip())
        emails.append(match.group(2))
        domains.append(match.group(3))
        match_lines.append(i)
df = pd.DataFrame()
df['Name'] = names
df['Email'] = emails
df['Domain'] = domains
print('Number of matching records:', len(df))
df.head(21)
```

```python
date_pattern = r'^Date:\s*(?:(\w{3}),\s*)?(\d{1,2}\s+\w{3}\s+\d{4})\s+(\d{1,2}:\d{2}:\d{2})'
days = []
dates = []
times = []
for i in range(len(match_lines)):
    from_line = match_lines[i]
    found = None
    for offset in range(1, 6):
        after = from_line + offset
        before = from_line - offset
        if after < len(lines):
            found = re.search(date_pattern, lines[after])
            if found:
                break
        if before >= 0:
            found = re.search(date_pattern, lines[before])
            if found:
                break
    if found:
        days.append(found.group(1))
        dates.append(found.group(2))
        times.append(found.group(3))
    else:
        days.append(None)
        dates.append(None)
        times.append(None)
df['Day'] = days
df['Date'] = dates
df['Time'] = times
df.head(21)
```
