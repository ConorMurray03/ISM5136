```python
import pandas as pd
df_insurance = pd.read_csv('auto_insurance_python.csv')
print("Shape before dropping:", df_insurance.shape)
mask = df_insurance['State'].notnull()
df_insurance = df_insurance[mask].copy()
print("Shape after dropping:", df_insurance.shape)
print("Missing State values remaining:", df_insurance['State'].isnull().sum())
```

```python
avg_income = df_insurance['Income'].mean()
print("Average Income:", round(avg_income, 2))
```

```python
states = ['Arizona', 'Oregon']
for i in range(len(states)):
    state_name = states[i]
    mask = df_insurance['State'] == state_name
    avg_policies = df_insurance[mask]['Number of Policies'].mean()
    print(state_name, "average Number of Policies:", round(avg_policies, 2))
```

```python
zero_mask = df_insurance['Income'] == 0
num_zero = zero_mask.sum()
total = len(df_insurance)
pct_zero = num_zero / total * 100
print("Customers with Income = 0:", num_zero)
print("Total customers:", total)
print("Percentage with Income = 0:", round(pct_zero, 2), "%")
nonzero_mask = df_insurance['Income'] != 0
print("Smallest non-zero Income:", df_insurance[nonzero_mask]['Income'].min())
```

```python
original = df_insurance['Number of Policies'].copy()
binned = original.copy()
mask_3 = original > 5
mask_2 = (original > 1) & (original <= 5)
mask_1 = original == 1
binned.loc[mask_3] = 3
binned.loc[mask_2] = 2
binned.loc[mask_1] = 1
df_insurance['Number of Policies'] = binned
print(df_insurance['Number of Policies'].value_counts().sort_index())
print("Data type:", df_insurance['Number of Policies'].dtype)
```

```python
df_insurance = df_insurance.drop(columns=['Customer ID'])
print(df_insurance.columns)
```

```python
df_insurance = pd.get_dummies(df_insurance,
                              columns=['State', 'Coverage', 'Vehicle Class'],
                              drop_first=True,
                              dtype=int)

print(df_insurance.shape)
print(df_insurance.columns)
df_insurance.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
cols_to_scale = ['Income', 'Months Since Policy Inception', 'Customer Lifetime Revenue']
df_insurance[cols_to_scale] = scaler.fit_transform(df_insurance[cols_to_scale])
print(df_insurance[cols_to_scale].describe().round(3))
df_insurance.head()
```

```python
print(df_insurance['Number of Policies'].value_counts().sort_index())
```

```python
max_income = df_insurance['Income'].max()
print("Max standardized Income:", round(max_income, 3))
```

```python
print("Number of attributes:", df_insurance.shape[1])
```

```python
mask_missing = df_insurance['Response'].isnull()
df_prediction = df_insurance[mask_missing].copy()

print("df_prediction shape:", df_prediction.shape)
print("Missing Response values in df_prediction:", df_prediction['Response'].isnull().sum())
df_prediction.head()
```

```python
mask_has_response = df_insurance['Response'].notnull()
df_insurance = df_insurance[mask_has_response]

print("df_insurance shape:", df_insurance.shape)
print("Missing Response values in df_insurance:", df_insurance['Response'].isnull().sum())
```

```python
print("Observations in df_insurance:", len(df_insurance))
print("Observations in df_prediction:", len(df_prediction))
```

```python
X = df_insurance.drop(columns=['Response'])
y = df_insurance['Response']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.neural_network import MLPClassifier
nn_model = MLPClassifier(hidden_layer_sizes=(5,), random_state=1)
nn_model = nn_model.fit(X_train, y_train)
print("Iterations used in training:", nn_model.n_iter_)
```

```python
y_pred = nn_model.predict(X_test)
print("First 10 predictions:", y_pred[0:10])
print(pd.Series(y_pred).value_counts())
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, pos_label=1)
recall = recall_score(y_test, y_pred, pos_label=1)
print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
tn = cm[0][0]
fp = cm[0][1]
fn = cm[1][0]
tp = cm[1][1]
print("True Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)
```

```python
X_prediction = df_prediction.drop(columns=['Response'])
predictions = nn_model.predict(X_prediction)
df_prediction['Response'] = predictions
print("Predicted Responses:", predictions)
print(df_prediction['Response'].value_counts())
df_prediction
```

```python
num_accept = (df_prediction['Response'] == 1).sum()
num_reject = (df_prediction['Response'] == 0).sum()
print("Predicted to accept (1):", num_accept)
print("Predicted to reject (0):", num_reject)
```

```python
X_prediction = df_prediction.drop(columns=['Response'])
print("Class order:", nn_model.classes_)
all_probs = nn_model.predict_proba(X_prediction)
prob_class1 = all_probs[:, 1]
for i in range(len(prob_class1)):
    print("Observation", df_prediction.index[i], "- P(Class=1):", round(prob_class1[i], 4))
```

```python
cutoff = 0.6
count_class1 = 0
for i in range(len(prob_class1)):
    if prob_class1[i] >= cutoff:
        count_class1 = count_class1 + 1
print("Observations classified as Class=1 at cutoff 0.6:", count_class1)
```

```python
cutoff = 0.6
count_accept = 0
for i in range(len(prob_class1)):
    if prob_class1[i] >= cutoff:
        count_accept = count_accept + 1
print("Predicted to accept at cutoff 0.6:", count_accept)
print("Predicted to reject at cutoff 0.6:", len(prob_class1) - count_accept)
```

```python
from sklearn.metrics import precision_score, recall_score
test_probs = nn_model.predict_proba(X_test)[:, 1]
cutoffs = [0.5, 0.6]
for i in range(len(cutoffs)):
    c = cutoffs[i]
    test_pred = []
    for j in range(len(test_probs)):
        if test_probs[j] >= c:
            test_pred.append(1)
        else:
            test_pred.append(0)
    print("Cutoff", c,
          "| Predicted 1s:", sum(test_pred),
          "| Precision:", round(precision_score(y_test, test_pred), 4),
          "| Recall:", round(recall_score(y_test, test_pred), 4))
```
